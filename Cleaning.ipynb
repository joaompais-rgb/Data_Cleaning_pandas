```python
def clean_column_names(df):
    df = df.copy()
    df.columns = df.columns.str.strip()
    df = df.rename(columns={"Fatal Y/N": "Fatal", "Fatal (Y/N)": "Fatal"})
    return df

new_df_shark_attacks = clean_column_names(df_shark_attacks)
list(new_df_shark_attacks.columns)
```

```python
columns_to_keep = [
    "Date", "Year", "Type", "Country", "State", "Location",
    "Activity", "Sex", "Age", "Injury", "Fatal", "Time", "Species",
]
new_df_shark_attacks = new_df_shark_attacks[columns_to_keep].copy()
new_df_shark_attacks.shape
```

```python
before = len(new_df_shark_attacks)
new_df_shark_attacks = new_df_shark_attacks.dropna(how="all")
print("Filas borradas:", before - len(new_df_shark_attacks))
```
