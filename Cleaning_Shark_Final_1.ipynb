```python
import re

MESES = {'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6,
         'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12}

def sacar_mes(valor):
    texto = str(valor).lower()
    encontrado = re.search(
        r'(jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)', texto)
    if encontrado:
        return MESES[encontrado.group(1)]
    return None

new_df_shark_attacks['month'] = (new_df_shark_attacks['month']
                                 .apply(sacar_mes).astype('Int64'))

new_df_shark_attacks['month'].value_counts(dropna=False).sort_index()
```

sacamos el número del mes de una columna que todavía guarda la fecha entera.

Por qué no usamos pd.to_datetime directamente: parece el camino obvio y es una trampa. El archivo mezcla fechas completas ("15-Jun-2024"), textos con prefijo ("Reported 10-Oct-2023") y valores que solo traen el año ("1903"). Un valor con solo el año se convertiría en el 1 de enero, y estaríamos inventando cientos de ataques en enero. El gráfico saldría precioso y estaría mal.

Por qué en minúsculas: str(valor).lower() hace que "Jun", "JUN" y "jun" encuentren el mismo patrón, sin tener que escribir las tres variantes.

NOTA:la cuenta de ataques por mes y cuántos quedan sin mes. Ese número va a las conclusiones, porque el análisis por estación se basa solo en los que sí lo tienen.

```python
HEMISFERIO_SUR = ['AUSTRALIA', 'SOUTH AFRICA', 'NEW ZEALAND', 'BRAZIL',
                  'ARGENTINA', 'CHILE', 'PERU', 'URUGUAY', 'FIJI',
                  'INDONESIA', 'PAPUA NEW GUINEA', 'MOZAMBIQUE',
                  'NAMIBIA', 'REUNION', 'REUNION ISLAND', 'MAURITIUS',
                  'NEW CALEDONIA', 'FRENCH POLYNESIA', 'MADAGASCAR',
                  'VANUATU', 'TONGA', 'SAMOA', 'AMERICAN SAMOA',
                  'SOLOMON ISLANDS']

OPUESTO = {'invierno': 'verano', 'verano': 'invierno',
           'primavera': 'otoño', 'otoño': 'primavera'}

def sacar_estacion(fila):
    mes = fila['month']
    if pd.isna(mes):
        return 'unknown'
    if mes in (12, 1, 2):
        estacion = 'invierno'
    elif mes in (3, 4, 5):
        estacion = 'primavera'
    elif mes in (6, 7, 8):
        estacion = 'verano'
    else:
        estacion = 'otoño'
    # country todavía no está limpia: "Australia" y "AUSTRALIA " tienen que contar igual
    pais = str(fila['country']).strip().upper().replace('?', '').strip()
    if pais in HEMISFERIO_SUR:
        estacion = OPUESTO[estacion]
    return estacion

new_df_shark_attacks['season'] = new_df_shark_attacks.apply(sacar_estacion, axis=1)

# comprobación: el mismo país escrito de dos maneras
print(new_df_shark_attacks['country'].str.strip().value_counts().filter(regex='(?i)australia'))
new_df_shark_attacks['season'].value_counts(dropna=False)
```

Convertimos el mes en estación, y la invertimos en los países del hemisferio sur.

**Por qué el hemisferio es imprescindible aquí:** junio es verano en Estados Unidos e invierno en Australia. Como Australia y Sudáfrica están entre los países con más ataques, sin corregirlo la mitad de nuestros datos quedaría con la estación cambiada, y la conclusión sobre cuándo reforzar la vigilancia sería exactamente la contraria de la verdadera.

**La función por dentro, y por qué `axis=1`:** hasta ahora `apply` trabajaba sobre una columna y recibía un valor. Aquí lo aplicamos sobre la tabla entera con `axis=1`, y entonces recibe una fila completa, de la que podemos leer el mes y el país a la vez. Sin `axis=1` recorrería columnas en vez de filas.

**Por qué ponemos el país en mayúsculas dentro de la función:** esta celda va antes de limpiar `country`, y el archivo escribe el mismo país de dos maneras: los registros antiguos en mayúsculas ("AUSTRALIA") y los recientes no ("Australia"). La lista está en mayúsculas, así que sin el `upper()` los ataques recientes de Australia no se invertían: la primera fila del archivo, Perth en septiembre, salía como otoño cuando allí es primavera. La comprobación de la salida enseña cuántas filas hay escritas de cada forma.

**Por qué la lista es más larga que antes:** faltaban Nueva Caledonia y la Polinesia Francesa, que están entre los países con más ataques y en el hemisferio sur, y otras islas del Pacífico sur y del Índico.

**Qué leemos:** la comprobación encuentra 1480 filas de Australia escritas «AUSTRALIA» y 56 escritas «Australia»: esas 56 son las que la versión anterior no invertía. En la tabla completa, todavía sin filtrar por años, salen 2684 ataques en verano, 1686 en otoño, 1313 en primavera y 838 en invierno, y 596 sin mes, que quedan como `unknown`.

### El año: número, y solo años completos

**Qué hacemos:** convertimos `year` en número entero y nos quedamos con los años de nuestros tres bloques, de 1951 a 2025.

**Las funciones por dentro:** `pd.to_numeric(columna, errors='coerce')` intenta convertir cada valor en número y, gracias a `errors='coerce'`, pone vacío en lo que no pueda convertir en vez de parar con un error. `astype('Int64')`, con I mayúscula, es el entero de pandas que acepta vacíos; el `int` normal daría error mientras quede uno. `between(ANIO_MIN, ANIO_MAX)` devuelve `True` para los años dentro del intervalo, con los dos extremos incluidos.

**Por qué número y no texto:** como texto, `max()` compara letras y no cantidades, y un filtro como `>= 1951` da error. La evolución por años necesita número.

**Por qué el máximo es 2025 y no el de los datos:** el archivo llega a septiembre de 2026, y un año sin octubre, noviembre y diciembre no se puede comparar con uno completo. Además el archivo se actualiza cada semana: si calculáramos el máximo con `max()`, en enero saldría 2027 y todos los números de la presentación cambiarían solos. Un número fijo hace el notebook reproducible.

**Por qué el mínimo es 1951:** no lo elegimos, sale del máximo. Tres bloques de 25 años que terminan en 2025 empiezan en 2025 − 74 = 1951.

**Qué leemos en la salida:** cuántas filas había y cuántas quedan, y el año mínimo y el máximo. Tienen que salir 1951 y 2025: ese print es la prueba de que el filtro hizo lo que decimos.

```python
ANIO_MAX = 2025                # último año completo: 2026 va por septiembre
ANIO_MIN = ANIO_MAX - 74       # 1951: tres bloques exactos de 25 años

new_df_shark_attacks['year'] = pd.to_numeric(
    new_df_shark_attacks['year'], errors='coerce').astype('Int64')

antes = len(new_df_shark_attacks)
new_df_shark_attacks = new_df_shark_attacks[
    new_df_shark_attacks['year'].between(ANIO_MIN, ANIO_MAX)]

print('Antes:', antes, '| Después:', len(new_df_shark_attacks))
print('Año mínimo:', new_df_shark_attacks['year'].min(),
      '| Año máximo:', new_df_shark_attacks['year'].max())
```

```python
def agrupar_años(year):
    if pd.isna(year):
        return None
    if year > ANIO_MAX or year < ANIO_MIN:
        return None                                  # fuera de los tres bloques
    if year >= ANIO_MAX - 24:
        return f'{ANIO_MAX - 24}-{ANIO_MAX}'         # 2001-2025
    elif year >= ANIO_MAX - 49:
        return f'{ANIO_MAX - 49}-{ANIO_MAX - 25}'    # 1976-2000
    else:
        return f'{ANIO_MAX - 74}-{ANIO_MAX - 50}'    # 1951-1975
```

```python
new_df_shark_attacks['bloque_año'] = new_df_shark_attacks['year'].apply(agrupar_años)
BLOQUE_RECIENTE = f'{ANIO_MAX - 24}-{ANIO_MAX}'

new_df_shark_attacks['bloque_año'].value_counts(dropna=False).sort_index()
```

**Qué hacemos:** partimos los años en tres bloques de 25 y guardamos el resultado en una columna nueva, `bloque_año`.

**La función por dentro:** `agrupar_años` recibe un año y devuelve el nombre de su bloque. Las preguntas van de la más reciente a la más antigua: el primer `if` que se cumple gana, así que un año de 2010 entra en el primero y ya no llega a los demás. Los bordes no están escritos a mano: salen de `ANIO_MAX`, y si algún día se cambia el año máximo, los tres bloques se mueven solos. `apply` aplica la función a cada año de la columna, uno a uno.

**Por qué bloques iguales:** para comparar tramos de tiempo, los tramos tienen que medir lo mismo. Un bloque de 25 años al lado de uno de 2 no dice nada.

**Por qué `BLOQUE_RECIENTE`:** es el nombre del bloque 2001-2025 guardado en una variable. El análisis lo usa en varias celdas, y así nadie tiene que escribir el texto a mano ni equivocarse con un guion.

**Qué leemos:** 1262 ataques en 1951-1975, 1223 en 1976-2000 y 2714 en 2001-2025, y ningún `None`: el filtro de arriba ya dejó solo años de los tres bloques. El último bloque tiene más del doble que cada uno de los anteriores, y eso es sobre todo registro.

NOTA: el recuento sube muchísimo hacia el presente. Eso **no** significa que haya más tiburones: significa que hoy se registra mejor. Los recuentos entre bloques no son comparables; la mortalidad, que es un porcentaje dentro de cada bloque, sí.

```python
def clasificar_actividad(texto):
    if pd.isna(texto):
        return 'unknown'
    texto = str(texto).lower()
    if 'surf' in texto or 'boogie' in texto or 'body board' in texto \
            or 'bodyboard' in texto or 'paddle' in texto or 'kite' in texto:
        return 'surfing'
    elif 'swim' in texto or 'bath' in texto or 'wading' in texto \
            or 'standing' in texto or 'floating' in texto:
        return 'swimming'
    elif 'fish' in texto:
        return 'fishing'
    elif 'div' in texto or 'snorkel' in texto or 'scuba' in texto:
        return 'diving'
    else:
        return 'other'
```

```python
def sacar_hora(texto):
    # «1015hrs», «0945hrs», «1635 hrs»: las dos primeras cifras son la hora
    encontrado = re.search(r'\b(\d{1,2})(\d{2})\s*hrs?\b', texto)
    if encontrado:
        return int(encontrado.group(1))
    # «14h00», «11h30», «14h»: la hora va delante de la h
    encontrado = re.search(r'\b(\d{1,2})\s*h', texto)
    if encontrado:
        return int(encontrado.group(1))
    # «10:30»
    encontrado = re.search(r'\b(\d{1,2}):\d{2}', texto)
    if encontrado:
        return int(encontrado.group(1))
    return None

def sacar_franja(valor):
    texto = str(valor).lower()
    hora = sacar_hora(texto)
    if hora is not None:
        if hora <= 5:
            return 'night'
        elif hora <= 11:
            return 'morning'
        elif hora <= 17:
            return 'afternoon'
        elif hora <= 23:
            return 'evening'
        else:
            return 'unknown'
    if 'morning' in texto or 'dawn' in texto:
        return 'morning'
    elif 'afternoon' in texto or 'midday' in texto or 'noon' in texto:
        return 'afternoon'
    elif 'evening' in texto or 'dusk' in texto or 'sunset' in texto:
        return 'evening'
    elif 'night' in texto:
        return 'night'
    else:
        return 'unknown'

# prueba con ejemplos reales del archivo antes de fiarnos
for ejemplo in ['1015hrs', '0945hrs', '1100hrs', '1635 hrs', '14h00', '11h30', 'Late afternoon', 'Dusk']:
    print(ejemplo, '->', sacar_franja(ejemplo))

new_df_shark_attacks['time_group'] = new_df_shark_attacks['time'].apply(sacar_franja)
new_df_shark_attacks['time_group'].value_counts(dropna=False)
```

**El problema:** la columna mezcla horas escritas de varias formas ("14h00", "11h30", "1015hrs", "1635 hrs") con palabras ("Afternoon", "late afternoon", "Dusk"). No es ni número ni categoría, es las dos cosas a la vez.

**El fallo que tenía la versión anterior:** buscaba "uno o dos dígitos seguidos de h". En "1015hrs" lo primero que cumple eso es "15h", así que las 10:15 salían como tarde; "1100hrs" daba "00h" y salía como noche; "0945hrs" daba "45h" y acababa en unknown. El formato de cuatro cifras es el de los registros recientes, justo los que más pesan en el análisis.

**La función por dentro:** `sacar_hora` prueba tres formatos, del más concreto al más general. Primero cuatro cifras seguidas de "hrs": `(\d{1,2})(\d{2})` separa la hora de los minutos, y nos quedamos con la hora. Después una o dos cifras seguidas de h, como en "14h00". Y por último "10:30". `\b` marca el borde de una palabra, para no empezar a leer a mitad de un número. Si no hay ninguna hora, `sacar_franja` busca palabras clave.

**Por qué la franja y no la hora exacta:** con la hora exacta tendríamos 24 barras y casi ningún caso en cada una. Cuatro franjas se leen de un vistazo, y es lo que una escuela necesita para decidir a qué horas meter a los alumnos en el agua.

```python
resumen = (surf_mundo.groupby('country').agg(
        ataques=('country', 'size'),
        playas=('location', lambda s: s[s != 'nan'].nunique()),
        con_edad=('age', 'count'),
        edad_media=('age', 'mean'),
        con_fatality=('fatality', 'count'),
        mortales=('fatality', lambda s: (s == 'Y').sum()))
    .sort_values('ataques', ascending=False)
    .head(15))

resumen['mortalidad'] = (resumen['mortales'] / resumen['con_fatality'] * 100).round(1)
resumen.round(1)
```

Juntamos en una sola tabla el número de ataques, en cuántas playas distintas, la
edad media, cuántos tienen edad registrada y el porcentaje de casos mortales.

**La función por dentro, `agg` con nombres:** `agg(nombre=('columna', 'función'))` calcula varias
métricas de golpe y deja cada una con el nombre que le damos. Es mucho más legible que un `agg`
con una lista, donde después hay que adivinar qué columna es cuál.

**Las cinco métricas, y por qué cada una:**
- `size` cuenta **filas**, o sea ataques. No usamos `count`, que contaría valores rellenos de una
  columna y perdería los ataques con campos vacíos.
- `nunique` en `location` dice en cuántas playas distintas ocurrieron. Es la métrica que separa
  "país peligroso" de "playa peligrosa": un país con muchos ataques concentrados en una sola playa
  puede ser perfectamente seguro a 50 km de allí.
- `count` en `age` son las edades registradas. Va al lado de la media a propósito.
- `mean` en `age` es la edad media, calculada solo sobre las que existen.
- La mortalidad se calcula con un `lambda`: `(s == 'Y')` da una columna de `True` y `False`, y su
  media es directamente la proporción de mortales, porque `True` vale 1. Multiplicada por 100, es
  el porcentaje.
  
NOTA: ningún porcentaje sin su número de casos. Un país con 2
ataques y 1 mortal da un 50% de mortalidad y parecería el lugar más peligroso del mundo; con la
columna de ataques al lado se ve enseguida que no se puede concluir nada.

**Dos cambios respecto a la versión anterior:** la mortalidad se divide entre los ataques con desenlace registrado (`con_fatality`) y no entre todos, porque un vacío no es un superviviente. Y `playas` ya no cuenta `'nan'` como si fuera una playa más.

**Qué leemos:** Estados Unidos concentra 556 ataques en 287 playas distintas, con una mortalidad del 2,2%: el riesgo está repartido por todo el país, y por eso hay que bajar al nivel de estado y de playa. Reunión tiene 24 ataques, 9 de ellos mortales (37,5%): con más de diez casos, es el destino con la mortalidad más alta de la tabla.
